```python
# 8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
'''
8.3. This gives an UnboundLocalError. Because we're assigning to counter inside the function (counter = counter + 1)
Python treats counter as a local variable for the whole function. So when it tries to read counter on the right side before it's been locally assigned, it fails
it doesn't know to look at the global one.
'''
```

```python
# 8.3
counter = 0

def increment_counter():
    global counter  # Declare counter as global
    counter = counter + 1

increment_counter()
print(counter)
```

```python
# 8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
