```python
import pandas as pd
import numpy as np
```

```python
student1={
    's_name' :['Amit', 'Ajay', 'Ravi' , 'Reeta', 'Geeta'],
    's_age': [18,20,np.nan,20,None],
    's_marks':[20,30,np.nan,70,np.nan]
}
```

```python
df=pd.DataFrame(student1)
df
```

```python
# detect or see missing value
df.isnull()
```

```python
# total count missing value
df.isnull().sum()
```

```python
# delete null , this by default row delete
df.dropna()
```

```python
# for col axis 1, for row axis 0
df.dropna(axis=1)
```

```python
df['s_age']=df['s_age'].fillna(df['s_age'].mean())
df
```

```python
df['s_marks']=df['s_marks'].fillna(df['s_marks'].median())
df
```

```python
df['Gender']=['Male', 'Female', np.nan, 'Male',np.nan]
df
```
