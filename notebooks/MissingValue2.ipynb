```python
import pandas as pd
import numpy as np
```

```python
Employee={
    'e_name' :['Ramlal', 'AjayDevGun', 'Sallu' , 'Reeki', 'Geto', 'AAA', 'AAA','Ramlal'],
    'e_age': [18,20,np.nan,20,None,40,40,28],
    'e_Department': ['HR', 'CS', np.nan,'ME',None,'CS','CS','HR'],
    'e_salary':[20000,300000,np.nan,700000,np.nan,50000,50000,20000]
}
```

```python
df=pd.DataFrame(Employee)
df
```

```python
df['e_salary']=df['e_salary'].fillna(0)
df
```

```python
df['e_Department']=df['e_Department'].fillna('unknown')
df
```

```python
df['e_Department']=df['e_Department'].ffill()
df
```

```python
df['e_age']=df['e_age'].bfill()
df
```

```python
df.drop_duplicates(inplace=True)
```
