SQLIT

```python
import sqlite3
```

```python
conn=sqlite3.connect("studentrecord.db")
cursor=conn.cursor()
print("database creatted succesfully")
```

```python
cursor.execute("""
              CREATE TABLE IF NOT EXISTS result(S_ID INTEGER, S_NAME CHAR, MARKS INTEGER)""")
print("table created successfully")
```

```python
cursor.execute("INSERT INTO result VALUES(?,?,?)",(1,"AMIT",99))
cursor.execute("INSERT INTO result VALUES(?,?,?)",(2,"REETA",98))
cursor.execute("INSERT INTO result VALUES(?,?,?)",(3,"RAHUL",97))
conn.commit()
print("record inserted")
conn.close()
```

```python
conn=sqlite3.connect("studentrecord.db")
cursor=conn.cursor()
cursor.execute("SELECT * FROM result")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```
