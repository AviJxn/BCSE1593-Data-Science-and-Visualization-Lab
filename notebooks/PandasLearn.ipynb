```python
# create new columns
df['Department']=['BTech(EC)','BTech(CS)','BTeach(EE)','BTech','BTech(CSE)']
df
```

```python
import numpy as np
```

```python
df['Grade'] = np.where((df['StudentMarks'] >= 90) & (df['StudentMarks']<=100), 'A',
              np.where((df['StudentMarks'] <90) & (df['StudentMarks'] >=70), 'B',
              np.where((df['StudentMarks'] <70) & (df['StudentMarks'] >=60), 'C',
              np.where((df['StudentMarks'] <60) & (df['StudentMarks'] >=40), 'D',
              np.where((df['StudentMarks'] < 40) , 'E', 'F')))))
df
```
